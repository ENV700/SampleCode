# Reading GIS Datasets

## Installing Geospatial Packages
To read geospatial datasets (shapefiles and rasters) into your coding environment, you'll need the `geopandas` and `rioxarray` packages which are not yet installed in your ENV700 Python environment created early on in the class. To do this, we'll create a copy of our existing environment (in case our attempt fails) and add these packages to that. Here's how:
1. Open the MiniForge prompt app on your Virtual Machine
1. Type the following command to clone you existing `ENV700` environment to a new one called `ENV700_gis`:<br>
    ```bash
    conda create --clone env700 --name env700_gis -y
    ```
1. Now make this new environment the active one:<br>
    ```bash
    activate env700_gis
    ```
1. Next, install geopandas:<br>
    ```bash
    conda install -c conda-forge geopandas rioxarray -y
    ```
You now have a new kernel you can choose in VS Code that has the GeoPandas and Rasterio packages!

## Reading Shapefiles into a GeoDataframe
Now we'll examine how shapefiles can be read into your coding environment, creating a new spin on the dataframe: the *Geo*Dataframe.

First, be sure to select your new `ENV700_gis` kernel in running this notebook. 

````python
# Install packages
from pathlib import Path
import pandas as pd
import geopandas as gpd
import seaborn as sns
````

````python
# Set paths
project_fldr = Path.cwd().parent
raw_fldr = project_fldr/'data'/'raw'
processed_fldr = project_fldr/'data'/'processed'
counties_shp = raw_fldr/'ne_10m_admin_2_counties.zip'
````

````python
# Read the shapefile
counties_gdf = gpd.read_file(filename = counties_shp)
counties_gdf.head()
````

Notice the last column. This contains the geometry of the feature, which enable us to do geospatial analysis. 

````python
counties_gdf.columns
````

````python
#A quick plot of the dataframe
counties_gdf.plot(column='REGION',figsize=(10,10))
````

### A quick workflow
* Subset North Counties for North Carolina and display them, coloring counties by the area of land

````python
counties_gdf.query('REGION == "NC"').plot(column='AREA_SQKM')
````

## Raster data
We will now explore how to import and view properties of a Raster dataset using the `rioxarray` package. 
* Documentation on `rioxarray`: https://vylor.github.io/rioxarray/stable/index.html
* A great intro tutorial: https://earthdatascience.org/courses/use-data-open-source-python/intro-raster-data-python/fundamentals-raster-data/

````python
#Import the package
import rioxarray
````

````python
#Import the Durham elevation tif
durham_dem = rioxarray.open_rasterio(
    filename=raw_fldr/'DurhamElev.tif', 
    masked=True
    )
````

````python
#Display its properties
durham_dem
````

````python
#Import the raster
world_hillshade = rioxarray.open_rasterio(
    filename=raw_fldr/'NE2_50M_SR'/'NE2_50M_SR.tif', 
    )
````
