# FEMA API Explorer
This notebook examines the FEMA API for redacted claims can be used to extract specific records. 
* Documentation for the dataset: <https://www.fema.gov/openfema-data-page/nfip-redacted-claims-v3>  
* Docuement for the API: <https://www.fema.gov/about/openfema/api>
* The REST endpoint of the API:<https://www.fema.gov/api/open/v3/NfipClaims?>

Of note is that specifying the CSV format generates a downloadable file that has to be ingested via IO. The JSON format can be read directly. 

```python
# Packages
from pathlib import Path
import requests
import pandas as pd
```

https://www.fema.gov/api/open/v3/NfipClaims?$count=true&$filter=(state%20eq%20%27NC%27)

```python
# Build the request
base_url = "https://www.fema.gov/api/open/v3/NfipClaims?"
API_params = {
    '$count':'false',
    '$filter':'(state eq \'NC\')',
    '$select':'',
    '$skip':'0',
    '$top':'100'
}

# Execute the request
response_raw = requests.get(
    url=base_url, 
    params=API_params)

# Convert the returned JSON object to a dictionary
response_dict = response_raw.json()
```
